````python
# ---- 数据加载与长表构建（Gemini + Bagel 双模型 × QIB 主口径 + 官方 1–5 对照臂 + 旧 1–5 审计）----
import json
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

BASE = Path('../synth_v61_pilot')          # notebook 在 reviews/ 下运行
if not (BASE / 'questions.jsonl').exists():
    BASE = Path('/yzp/zhaozy/yangzepeng/0905/demiwtg/benchmark/edit/synth_v61_pilot')
EDIT_DATA = BASE.parent
PHI = {0: 0, 1: 60, 2: 100}
MODEL_COLOR = {'Gemini': '#d17a22', 'Bagel': '#2a6fbb'}
TYPE_ORDER = ['replace', 'add', 'adjust', 'remove', 'style',
              'action', 'extract', 'background', 'compose']
pd.set_option('display.max_columns', 60)
plt.rcParams['figure.dpi'] = 110


def _jsonl(p):
    p = Path(p)
    return [json.loads(l) for l in p.open(encoding='utf-8') if l.strip()] if p.exists() else []


def _json(p):
    p = Path(p)
    return json.loads(p.read_text(encoding='utf-8')) if p.exists() else None


QS = {q['qid']: q for q in _jsonl(BASE / 'questions.jsonl')}
GEM = {r['qid']: r for r in _jsonl(BASE / 'scores_qib/gemini/scores.jsonl')}
OLD = {r['qid']: r for r in _jsonl(BASE / 'scores/codex_blind/scores.jsonl')}
BAG = {r['qid']: r for r in _jsonl(BASE / 'scores_qib/bagel/scores.jsonl')}
GEM_REP = _json(BASE / 'scores_qib/gemini/report.json')
OLD_REP = _json(BASE / 'scores/codex_blind/report.json')
BAG_REP = _json(BASE / 'scores_qib/bagel/report.json')
PAIRED_REP = _json(BASE / 'scores_qib/paired/report.json')
PAIRED = {r['qid']: r for r in _jsonl(BASE / 'scores_qib/paired/paired_scores.jsonl')}
# ImgEdit 官方 1–5 对照臂（scores_official/ 与 QIB 主口径物理隔离，此处只读展示，禁换算互比）
OFF = {m: {r['qid']: r for r in _jsonl(BASE / f'scores_official/{m.lower()}/scores.jsonl')}
       for m in ('Gemini', 'Bagel')}
OFF_REPS = {m: _json(BASE / f'scores_official/{m.lower()}/report.json') for m in ('Gemini', 'Bagel')}
OFF_PAIRED_REP = _json(BASE / 'scores_official/paired/report.json')
OPDF = pd.DataFrame(_jsonl(BASE / 'scores_official/paired/paired_scores.jsonl')).set_index('qid')
# QIB v2.1 验枪轮（同 40 图·同判官·判准 v2.1；与 v2 冻结分禁直接互比，仅供第六节判准效应分析）
V21 = {m: {r['qid']: r for r in _jsonl(BASE / f'scores_qib_v21/{m.lower()}/scores.jsonl')}
       for m in ('Gemini', 'Bagel')}
V21_REPS = {m: _json(BASE / f'scores_qib_v21/{m.lower()}/report.json') for m in ('Gemini', 'Bagel')}
V21_PAIRED_REP = _json(BASE / 'scores_qib_v21/paired/report.json')
V21_PDF = pd.DataFrame(_jsonl(BASE / 'scores_qib_v21/paired/paired_scores.jsonl')).set_index('qid')
assert QS and len(GEM) == 20 and len(BAG) == 20 and len(PAIRED) == 20, '需两模型 20 题全部落盘'
assert len(OFF['Gemini']) == 20 and len(OFF['Bagel']) == 20 and len(OPDF) == 20, '需官方对照臂两模型 20 题全部落盘'
assert len(V21['Gemini']) == 20 and len(V21['Bagel']) == 20 and len(V21_PDF) == 20, '需 v2.1 验枪轮两模型 20 题全部落盘'


def _meta(qid):
    q = QS[qid]
    return dict(level=q.get('level', '?'), edit_type=q.get('edit_type', '?'),
                suite=q.get('suite', '?'), batch=q.get('_batch', '?'))


dim_rows, tot_rows = [], []
for model, scores in (('Gemini', GEM), ('Bagel', BAG)):
    for qid in sorted(scores):
        r = scores[qid]
        raw = [d['tier'] for d in r['raw_dimensions']]
        off = [raw[0], min(raw[1], raw[0]), min(raw[2], raw[0])]
        for i, d in enumerate(r['raw_dimensions']):
            dim_rows.append({**_meta(qid), 'qid': qid, 'model': model, 'dim': f'd{i + 1}',
                             'label': d['label'], 'tier': raw[i], 'mapped': d['mapped'],
                             'off_tier': off[i], 'off_mapped': PHI[off[i]],
                             'reason': d.get('reason', '')})
        tot_rows.append({**_meta(qid), 'qid': qid, 'model': model,
                         'official_total': r['official_total'],
                         'mapped_total': round(sum(d['mapped'] for d in r['raw_dimensions']) / 3, 3),
                         'raw_tier_mean': r.get('raw_tier_mean'),
                         'validity': r['validity']['status'],
                         'confidence': r.get('confidence')})

DIM = pd.DataFrame(dim_rows)          # 维度级长表: 2 模型 × 20 题 × 3 维 = 120 行
TOT = pd.DataFrame(tot_rows)          # 题目级: 2 模型 × 20 题 = 40 行

_gdir = next((BASE / 'gemini/imgs').glob('*'), None) if (BASE / 'gemini/imgs').exists() else None
IMG = {
    'before': lambda q: EDIT_DATA / q['_sample_image'],
    'gemini': lambda q: _gdir / f"{q['qid']}.png" if _gdir else None,
    'bagel': lambda q: BASE / 'bagel/imgs' / f"{q['qid']}.png",
}

gm = TOT[TOT.model == 'Gemini']
bm = TOT[TOT.model == 'Bagel']
print(f"题库 {len(QS)} 题 ｜ Gemini QIB {len(GEM)} ｜ Bagel QIB {len(BAG)}"
      f" ｜ 官方臂 G {len(OFF['Gemini'])} / B {len(OFF['Bagel'])}"
      f" ｜ Gemini 旧1-5(审计) {len(OLD)} ｜ paired QIB {len(PAIRED)} / 官方 {len(OPDF)}")
print(f"Gemini: QIB official {gm.official_total.mean():.2f} / mapped {gm.mapped_total.mean():.2f}"
      f"（report {GEM_REP['overall']}）｜官方臂 {OFF_REPS['Gemini']['overall']}/5")
print(f"Bagel : QIB official {bm.official_total.mean():.2f} / mapped {bm.mapped_total.mean():.2f}"
      f"（report {BAG_REP['overall']}）｜官方臂 {OFF_REPS['Bagel']['overall']}/5")
_w21 = V21_PAIRED_REP['overall']['wins_ties_losses']
for _m in ('Gemini', 'Bagel'):
    assert V21_REPS[_m]['n_invalid'] == 0 and all(
        r['judge'] == 'gpt-5.6-sol-built-in' and r['schema'] == 'edit-codex-v2-qib'
        for r in V21[_m].values()), f'v2.1 轮 {_m} 验收失败'
print(f"QIB v2.1 验枪轮: G {V21_REPS['Gemini']['overall']} / B {V21_REPS['Bagel']['overall']}"
      f"（report 对账一致；W/T/L {_w21['win']}/{_w21['tie']}/{_w21['loss']}"
      f"，Δ {V21_PAIRED_REP['overall']['delta_left_minus_right']}）")

def _overall_score_change_md(base):
    """同一批候选的三版 official 均分、升降分值与升降题数。"""
    from decimal import Decimal
    score_sets = {}
    means = {}
    for version, folder in [('V2.0', 'scores_qib'), ('V2.1', 'scores_qib_v21'), ('V2.2', 'scores_qib_v22')]:
        score_sets[version] = {}
        means[version] = {}
        for model in ('Gemini', 'Bagel'):
            rows = _jsonl(Path(base) / folder / model.lower() / 'scores.jsonl')
            assert len(rows) == 20 and all(r['validity']['status'] == 'ok' for r in rows), '本表按三版同批 20 题比较'
            values = {r['qid']: Decimal(str(r['official_total'])) for r in rows}
            score_sets[version][model] = values
            means[version][model] = sum(values.values()) / len(values)
            report = _json(Path(base) / folder / model.lower() / 'report.json')
            assert round(means[version][model], 3) == Decimal(str(report['overall']))
        means[version]['合并（40 个候选）'] = (means[version]['Gemini'] + means[version]['Bagel']) / 2
    for model in ('Gemini', 'Bagel'):
        assert set(score_sets['V2.0'][model]) == set(score_sets['V2.1'][model]) == set(score_sets['V2.2'][model])

    def change(value):
        return ('↑ +' if value > 0 else '↓ −' if value < 0 else '持平 ') + f'{abs(value):.3f}'

    combined21 = means['V2.1']['合并（40 个候选）']
    combined22 = means['V2.2']['合并（40 个候选）']
    rows = ['## 整体得分变化\n',
            '**V2.2 相对 V2.1，两模型得分均回升；相对 V2.0，两模型仍偏低。**',
            f'40 个候选的合并均分从 **{combined21:.3f} 升至 {combined22:.3f}，增加 {combined22-combined21:.3f} 分**。\n',
            '| 候选模型 | V2.0 均分 | V2.1 均分 | V2.2 均分 | V2.2 − V2.1 | V2.2 − V2.0 |',
            '|---|---:|---:|---:|---:|---:|']
    for model in ('Gemini', 'Bagel', '合并（40 个候选）'):
        a, b, c = (means[v][model] for v in ('V2.0', 'V2.1', 'V2.2'))
        rows.append(f'| {model} | {a:.3f} | {b:.3f} | {c:.3f} | {change(c-b)} | {change(c-a)} |')
    counts = []
    total_up = total_down = total_same = 0
    for model in ('Gemini', 'Bagel'):
        previous, current = score_sets['V2.1'][model], score_sets['V2.2'][model]
        up = sum(current[q] > previous[q] for q in current)
        down = sum(current[q] < previous[q] for q in current)
        same = len(current) - up - down
        counts.append(f'{model}：{up} 题上升、{down} 题下降、{same} 题不变')
        total_up += up; total_down += down; total_same += same
    rows.extend(['\n**V2.1 → V2.2 的逐题总分变化**：' + '；'.join(counts) + f'。合计 {total_up} 个候选上升、{total_down} 个下降、{total_same} 个不变。',
                 '\n口径：均为 official 钳后分（0–100），变化量单位为“分”。每个模型 20 题，合并行为 40 个候选的均值，等价于两模型等权均值。分数升降反映本轮判准及判官执行的变化，不等同于模型能力变化或评分准确率提升。'])
    return '\n'.join(rows) + '\n'
````

## 一、整体分布分析（两模型 × QIB 两口径 ＋ 官方对照臂）

总览：题目级 official/mapped 均值、钳制损失、0 分卷数；主维 d1（任务成立）三档题数；以及 official_total 的离散分布（钳制口径下每题得分只能落在少量离散值上）。

## 四、两种打分口径分析

口径 A（official 钳制）是横向比较主口径；口径 B（mapped 未钳制）反映各维原始能力。此处按 edit_type 对比两口径下的模型差与排名，检查**钳制是否改变结论方向**。附录为 Gemini 旧 1–5 诊断分与 QIB 的一致性审计（仅 Gemini、已废弃口径，禁止换算混用）。

````python
# ---- 四、双口径：edit_type 级排名与方向一致性 ----
rows = []
for t9 in [x for x in TYPE_ORDER if x in set(TOT.edit_type)]:
    sub = TOT[TOT.edit_type == t9]
    row = {'edit_type': t9}
    for model, tag in (('Gemini', 'G'), ('Bagel', 'B')):
        s = sub[sub.model == model]
        row[f'{tag}_official'] = round(s.official_total.mean(), 1)
        row[f'{tag}_mapped'] = round(s.mapped_total.mean(), 1)
        row[f'{tag}_钳制损失'] = round(s.mapped_total.mean() - s.official_total.mean(), 1)
    row['Δofficial(G-B)'] = round(row['G_official'] - row['B_official'], 1)
    row['Δmapped(G-B)'] = round(row['G_mapped'] - row['B_mapped'], 1)
    row['方向一致'] = ('一致' if np.sign(row['Δofficial(G-B)']) == np.sign(row['Δmapped(G-B)'])
                   else '翻转!')
    rows.append(row)
df = pd.DataFrame(rows).set_index('edit_type')
for m in ('G', 'B'):
    df[f'{m}排名official'] = df[f'{m}_official'].rank(ascending=False).astype(int)
    df[f'{m}排名mapped'] = df[f'{m}_mapped'].rank(ascending=False).astype(int)
    df[f'{m}排名变动'] = df[f'{m}排名mapped'] - df[f'{m}排名official']
display(df)

flip = df[df['方向一致'] == '翻转!']
print('两口径下 G-B 方向翻转的类型：', list(flip.index) if len(flip) else '无')
````

## 五、对照臂：ImgEdit 官方 rubric（1–5）三口径对比

同一批 40 张图，官方臂采用 gpt-5.6-sol；QIB V2.0 的 Bagel 组实际采用 gpt-6-astra，因此该组同时更换了判官模型，不能视为只改 rubric：ImgEdit 官方 1–5 逐字臂 vs QIB 0/60/100 主口径。注意两臂差异不止刻度——还有输出负担（结构化 JSON 逐维证据 vs ≤20 词推理）、钳制可见性（QIB 判官不见钳制 vs 官方文本内含钳制句）、锚点措辞（QIB 吸收了"记 0 须可指认证据、Excel 禁加戏"等证据规则 vs 官方原文逐字）、判据语言（中 vs 英）。因此本节展示的是**判分协议与执行条件共同变化的结果**，不是纯刻度消融。线性归一（(x−1)/4 → %）仅作松紧示意，不构成分数换算；对照臂结果不进入主口径正式结论。

````python
# ---- 五、三口径对比：总分与逐题一致性 ----
# OFF / OFF_REPS / OFF_PAIRED_REP / OPDF：官方对照臂数据已在加载格（cell 1）统一读入

rows = []
for model in ('Gemini', 'Bagel'):
    t = TOT[TOT.model == model]
    o = OFF_REPS[model]['overall']
    rows.append({'model': model, 'official 1-5': o,
                 'official线性归一%': round((o - 1) / 4 * 100, 1),
                 'QIB_official%': round(t.official_total.mean(), 1),
                 'QIB_mapped%': round(t.mapped_total.mean(), 1)})
summ = pd.DataFrame(rows).set_index('model')
summ['松紧差(official归一-QIB official)'] = summ['official线性归一%'] - summ['QIB_official%']
display(summ)

rows = []
for qid in sorted(QS):
    row = {'qid': qid, 'edit_type': QS[qid]['edit_type']}
    for model, tag in (('Gemini', 'G'), ('Bagel', 'B')):
        row[f'{tag}_official15'] = OFF[model][qid]['official_total']
        row[f'{tag}_QIB'] = float(TOT[(TOT.qid == qid) & (TOT.model == model)].official_total.iloc[0])
        row[f'{tag}_offd1'] = OFF[model][qid]['raw_dimensions'][0]['score']
        row[f'{tag}_qibfail'] = row[f'{tag}_QIB'] == 0
    rows.append(row)
X = pd.DataFrame(rows).set_index('qid')
# 旧 1–5 审计分（仅 Gemini 的 v1 废弃口径；与对照臂/QIB 只并列列示，禁线性互换）
X['G_old15_raw'] = [OLD[q]['raw_total'] for q in X.index]
X['G_old15_d1'] = [OLD[q]['raw_dimensions'][0]['score'] for q in X.index]
rho_oo = X['G_old15_raw'].corr(X['G_official15'], method='spearman')
a_old, a_off = X['G_old15_d1'] <= 2, X['G_offd1'] <= 2
print(f'Gemini 旧审计臂 vs 官方对照臂（两代 1–5，仅参照）：Spearman={rho_oo:.3f}'
      f' ｜ Fail判定一致率(各自 d1≤2) {(a_old == a_off).mean():.0%}'
      f'（双Fail {int((a_old & a_off).sum())} / 仅官方判Fail {int((~a_old & a_off).sum())} / 仅旧臂判Fail {int((a_old & ~a_off).sum())}）')
for model, tag in (('Gemini', 'G'), ('Bagel', 'B')):
    rho = X[f'{tag}_official15'].corr(X[f'{tag}_QIB'], method='spearman')
    a = X[f'{tag}_offd1'] <= 2      # official 臂：第一维 ≤2 ≈ 任务未成立
    b = X[f'{tag}_qibfail']         # QIB 臂：official_total==0 ⇔ d1 tier 0
    print(f"{model}: Spearman(official,QIB)={rho:.3f} ｜ Fail判定一致率(official d1≤2 vs QIB d1=0) "
          f"{(a == b).mean():.0%}（双Fail {int((a & b).sum())} / 双非Fail {int((~a & ~b).sum())} / "
          f"仅official判Fail {int((a & ~b).sum())} / 仅QIB判Fail {int((~a & b).sum())}）")

w1 = PAIRED_REP['overall']['wins_ties_losses']
o2 = OFF_PAIRED_REP['overall']; w2 = o2['wins_ties_losses']
print(f"\nW/T/L(G视角): QIB = {w1['win']}/{w1['tie']}/{w1['loss']} → official臂 = {w2['win']}/{w2['tie']}/{w2['loss']}"
      f" ｜ Δ(G-B): QIB {PAIRED_REP['overall']['delta_left_minus_right']} vs official {o2['delta_left_minus_right']}")
cmpw = pd.DataFrame({'QIB_winner': PDF.set_index('qid').winner, 'official_winner': OPDF.winner,
                     'QIB_Δ': PDF.set_index('qid').delta_left_minus_right,
                     'official_Δ': OPDF.delta_left_minus_right})
cmpw['edit_type'] = [QS[q]['edit_type'] for q in cmpw.index]
display(cmpw.T)
flip = cmpw[cmpw.QIB_winner != cmpw.official_winner]
print('胜者翻转题：', [f"{q}({r.edit_type}) {r.QIB_winner}→{r.official_winner}" for q, r in flip.iterrows()] or '无')
````

````python
# ---- 六、验枪总览：四臂总分 + 三轮配对 + 钳制损失 ----
rows = []
for model in ('Gemini', 'Bagel'):
    src2 = GEM if model == 'Gemini' else BAG
    qib2_rep = GEM_REP if model == 'Gemini' else BAG_REP
    v2_mapped = round(sum(sum(d['mapped'] for d in r['raw_dimensions']) / 3 for r in src2.values()) / 20, 2)
    v21_mapped = round(sum(sum(d['mapped'] for d in r['raw_dimensions']) / 3 for r in V21[model].values()) / 20, 2)
    rows.append({
        'model': model,
        'QIBv2_official钳(冻结)': qib2_rep['overall'],
        'QIBv2_mapped未钳': v2_mapped,
        'QIBv2.1_official钳(验枪)': V21_REPS[model]['overall'],
        'QIBv2.1_mapped未钳': v21_mapped,
        '判准效应Δ钳后(v2.1-v2)': round(V21_REPS[model]['overall'] - qib2_rep['overall'], 3),
        'v2.1钳制损失': round(v21_mapped - V21_REPS[model]['overall'], 2),
        '官方臂1-5': OFF_REPS[model]['overall'],
        '旧1-5审计': OLD_REP['overall'] if model == 'Gemini' else None,
    })
display(pd.DataFrame(rows).set_index('model'))
print('各列刻度不同（0/60/100 与 1–5），只并列不换算；旧 1–5 仅 Gemini 审计轮，Bagel 无该轮。')

w2 = PAIRED_REP['overall']['wins_ties_losses']
w21 = V21_PAIRED_REP['overall']['wins_ties_losses']
wo = OFF_PAIRED_REP['overall']['wins_ties_losses']
print(f"W/T/L(G视角)：QIB v2 = {w2['win']}/{w2['tie']}/{w2['loss']}（Δ {PAIRED_REP['overall']['delta_left_minus_right']}）"
      f" → QIB v2.1 = {w21['win']}/{w21['tie']}/{w21['loss']}（Δ {V21_PAIRED_REP['overall']['delta_left_minus_right']}）"
      f" ｜ 官方臂 = {wo['win']}/{wo['tie']}/{wo['loss']}（Δ {OFF_PAIRED_REP['overall']['delta_left_minus_right']}，1-5 刻度）")
# 对账：v2.1 report overall 与逐题重算一致
for model in ('Gemini', 'Bagel'):
    mine = round(sum(r['official_total'] for r in V21[model].values()) / 20, 3)
    assert abs(mine - V21_REPS[model]['overall']) < 1e-9, f'v2.1 {model} 对账不一致'
print('v2.1 report.json 对账：重算精确一致（G/B overall）')
````

````python
# ---- 六、v2→v2.1 逐题迁移：总分变化、d1 档位迁移、Fail 翻转 ----
rows = []
for qid in sorted(QS):
    row = {'qid': qid, 'edit_type': QS[qid]['edit_type']}
    for model, tag in (('Gemini', 'G'), ('Bagel', 'B')):
        src2 = GEM if tag == 'G' else BAG
        v2, v21 = src2[qid], V21[model][qid]
        row[f'{tag}_v2'] = v2['official_total']
        row[f'{tag}_v2.1'] = v21['official_total']
        row[f'{tag}_Δ'] = round(v21['official_total'] - v2['official_total'], 1)
        row[f'{tag}_d1迁移'] = f"{v2['raw_dimensions'][0]['tier']}→{v21['raw_dimensions'][0]['tier']}"
    rows.append(row)
SH = pd.DataFrame(rows).set_index('qid')
display(SH)

for model, tag in (('Gemini', 'G'), ('Bagel', 'B')):
    src2 = GEM if tag == 'G' else BAG
    v2 = [src2[q]['official_total'] for q in sorted(QS)]
    v21 = [V21[model][q]['official_total'] for q in sorted(QS)]
    rho = pd.Series(v2).corr(pd.Series(v21), method='spearman')
    f2 = [x == 0 for x in v2]; f21 = [x == 0 for x in v21]
    both = sum(a and b for a, b in zip(f2, f21)); only2 = sum(a and not b for a, b in zip(f2, f21))
    only21 = sum(b and not a for a, b in zip(f2, f21)); none = sum(not a and not b for a, b in zip(f2, f21))
    print(f"{model}: Spearman(v2, v2.1) = {rho:.3f} ｜ Fail(=0) 迁移：双Fail {both} / 仅v2Fail(升档) {only2}"
          f" / 仅v2.1Fail(新降) {only21} / 双非Fail {none}")

print('手册结论5 重点题（v2 QIB 满分 100 但官方臂仅 3.0/5，v2.1 Excel 判据改「全部精确+可核验」后预期回落）：')
for qid in ('e003', 'e018', 'e019'):
    r = SH.loc[qid]
    print(f"  {qid}({r['edit_type']}): G {r['G_v2']}→{r['G_v2.1']}（d1 {r['G_d1迁移']}）｜ B {r['B_v2']}→{r['B_v2.1']}")
````

````python
# 逐题合并视图：每个候选、版本、维度只展示一次评分理由。
import base64
import html
import io
import json
import re
from pathlib import Path
from functools import lru_cache
from IPython.display import display
from PIL import Image
import mistune

REVIEW_BASE = Path(globals().get('BASE', '../synth_v61_pilot')).resolve()
def _review_json(path):
    return json.loads(Path(path).read_text())
def _review_rows(path):
    return {r['qid']: r for r in map(json.loads, Path(path).read_text().splitlines())}
def _review_esc(text):
    return html.escape(str(text), quote=True)
def _review_num(number):
    return f'{number:.3f}'.rstrip('0').rstrip('.')
def _review_link(path, title):
    p = Path(path)
    relative = '../synth_v61_pilot/' + p.relative_to(REVIEW_BASE).as_posix()
    return f'<a href="{_review_esc(relative)}">{_review_esc(title)}</a>'
def _review_fold(title, body):
    return '<details><summary>' + _review_esc(title) + '</summary>' + body + '</details>'

REVIEW_RUNS = [
    ('V2.0', 'scores_qib'), ('V2.1', 'scores_qib_v21'),
    ('V2.2', 'scores_qib_v22'),
    ('V2.2 修订 · sol', 'scores_qib_v22_revision_20260907'),
    ('V2.2 修订 · sol high', 'scores_qib_v22_sol_high_20260907'),
    ('V2.2 修订 · astra', 'scores_qib_v22_astra_20260907'),
]
REVIEW_SCORES = {}
for _version, _folder in REVIEW_RUNS:
    for _candidate in ['Gemini', 'Bagel']:
        _path = REVIEW_BASE / _folder / _candidate.lower() / 'scores.jsonl'
        if _path.exists():
            REVIEW_SCORES[_candidate, _version] = _review_rows(_path)
REVIEW_INPUTS = {
    'Gemini': _review_rows(REVIEW_BASE / 'scores_qib_v22_astra_20260907/gemini/blind_manifest.jsonl'),
    'Bagel': _review_rows(REVIEW_BASE / 'scores_qib_v22_revision_20260907/bagel/blind_manifest.jsonl'),
}
REVIEW_QIDS = sorted(REVIEW_INPUTS['Gemini'])
assert len(REVIEW_QIDS) == 20
assert all(set(rows) == set(REVIEW_QIDS) for rows in REVIEW_SCORES.values())
REVIEW_PROVENANCE = _review_json(REVIEW_BASE / 'qib_judge_model_history_20260907.json')
REVIEW_JUDGES = {(r['candidate'], r['version']): r for r in REVIEW_PROVENANCE['summary']}
REVIEW_AUDIT = _review_json(REVIEW_BASE / 'scores_qib_v22_astra_20260907/audits/regression_review_after_scoring.json')['notes']
REVIEW_LEGACY = {
    'Gemini · 官方 1–5': _review_rows(REVIEW_BASE / 'scores_official/gemini/scores.jsonl'),
    'Bagel · 官方 1–5': _review_rows(REVIEW_BASE / 'scores_official/bagel/scores.jsonl'),
    'Gemini · 旧 1–5 审计': _review_rows(REVIEW_BASE / 'scores/codex_blind/scores.jsonl'),
}
_review_markdown = mistune.create_markdown(escape=False, plugins=['table'])

# EARLY_REVIEW_NOTES is inserted below when the notebook is consolidated.
EARLY_REVIEW_NOTES = {'e003': '### e003 / Gemini：评分后复核（原始评分保留）\n\n复核结论：V2.2 主维理由中的“大幅移位或尺度变化、明显不同取景”缺少图像证据。中央目标已完整消失，原已可见的左右挖掘机、卡车驾驶室、三根管道和路锥总体对应；卡车车厢露出更多是解除遮挡后的合理补全。\n\n两张实际评分输入的尺寸为 BEFORE 1672×941、AFTER 1376×768，宽高比分别为 1.777、1.792，相差约 0.83%。特征对应更接近约 0.823 倍的整体等比例缩放及约 3 像素的纵向平移，不能据此认定对象大幅重新布局。\n\n在五处原已可见的保留区域，匹配点的中位横向位置差绝对值均不超过画宽 0.03%，纵向不超过画高 0.36%。该数值按各自画幅归一化、在几何配准修正之前计算；SIFT 双向匹配和全局 RANSAC 仅用于剔除误匹配，不是对整幅图逐像素完全相同的保证。\n\nV2.2 的 remove / Prompt Compliance 2 档已经要求“保持比较限于原已可见部分，解除遮挡后的面积增加不等于移位，几何判断须以同尺度下的固定参照物为依据”。本次判官虽写了“相对画框”，却未提供成立的对应差异。此核验要求应同时约束 0 档的保持破坏判定，不能只在审查满档时执行。\n\n本次审阅倾向将主维评为 2；若沿用原二、三维的 2，则建议为 (2,2,2)，对应 100 分。这是评分后复核意见，不是新一轮盲评分；原始冻结记录仍保留 (0,2,2) 与总分 0，现有总体均分未调整。\n\n[几何核验记录](../synth_v61_pilot/scores_qib_v22/audits/e003_geometry_review_20260907.json)\n\n\n**sol 修订轮评分后复核：** 反馈回归通过：Gemini raw (0,2,2)→(2,2,2)，总分 0→100。正确对应左右挖掘机、自卸车、混凝土管及路锥，没有再凭尺寸差异虚构大幅移位。', 'e004': '### e004 / Gemini：评分后复核（原始评分保留）\n\n复核结论：原理由中的“京变成皖”和街景“大范围身份与结构改变”不能按原文成立。右前方牌内图形确已改为奔跑；行人、两把伞、车辆及交警的主要位置与形态总体对应，不能仅凭细节变化推出整片场景身份与结构被破坏。\n\nBEFORE 为 2752×1536，AFTER 为 1376×768，长宽各缩为一半，宽高比不变。六处保留区域的匹配点，在几何配准修正前按画幅归一化计算，中位横、纵位置差绝对值均不超过画幅 0.02%；这支持布局基本对应，但不是所有像素或面部细节完全一致的保证。\n\n车牌不能全部归因于分辨率：把 BEFORE 按面积平均降采样到 AFTER 的分辨率后，两车车牌的“京”仍基本可辨；AFTER 的首字存在额外笔画损失或字形变形。因此有局部文字保真损失，但可见笔画不足以确认具体改成了“皖”。应写“首字字形退化、无法可靠识别”，不能把不确定 OCR 写成已确认的省份替换。\n\n原主维 0 档理由须重新审查；有无局部保真损失、是否达到实质语义破坏，必须分别论证，不能从一个难辨字扩大为所有人物和车辆的身份破坏。第三维评价调整后对象的结构、材料及环境关系，原全景重绘叙述也没有给出足够独立的物理失败依据。本次不重新指定总分，原 (0,1,0) / 0 分及均分保留。\n\n下图三列依次是 BEFORE 原始像素、BEFORE 降采样至 AFTER 分辨率、AFTER 原始像素；只作对应裁切、面积降采样控制及最近邻显示，没有锐化、超分辨率或生成式补笔画。\n\n![车牌原始像素与同分辨率控制对照](../synth_v61_pilot/scores_qib_v22/audits/e004_plate_pixel_review_20260907.png)\n\n[几何与裁切核验记录](../synth_v61_pilot/scores_qib_v22/audits/e004_geometry_review_20260907.json)\n\n\n**sol 修订轮评分后复核：** 反馈回归未通过：Gemini raw (0,1,0)→(0,2,0)。虽不再猜“京→皖”，仍声称两块车牌字符、人物和街景广泛改变，并据此给主维和物理维 0；同尺度审计只支持部分小字可读性下降，不支持这种整体变化。Bagel 的 100 是另一候选，不能混为 Gemini 已修复。', 'e006': '### e006 / Gemini：评分后复核（原始评分保留）\n\n复核结论：Gemini 结果确实从六名变成七名工人。BEFORE 为左侧织机旁一名、货架区五名；AFTER 这六名仍可对应，红色消防箱旁另出现一名面向镜头、双臂做 X 手势的人。原最右侧背身工人仍在新增者左侧货架旁，不是原工人移动或转身。\n\n三版使用的源图、结果图和题面哈希完全相同，Gemini 三份评分均登记 gpt-5.6-sol-built-in。V2.0 原始 (0,0,1)、总分 0；V2.1 原始 (0,0,1)、总分 0，两版理由都准确指出六人变七人和新增人物。V2.2 原始 (1,1,2)、总分 60，误写“前后均可数出六名”，把新增者错误对应为指定工人。\n\n依 V2.2 自身规则也应失败：Action Fidelity 的 0 档明确包含“错主体、原动作仍清晰存在”；Identity Preservation 的 0 档要求核验“新增或缺失对象”。因此主维与身份保持均应为 0，钳后总分应为 0。第三维可独立认可新增人物的解剖合理性，无须为了与总分一致强行判成 0。本段是复核意见，冻结原始评分及均分未改。\n\n这里先发生了人数漏计和目标对应错误，不能解释成 V2.2 允许新增人物。文字上另有值得修正的边界：V2.1 的 1 档要求“动作的方向、幅度与主体姿态达到指令要求”；V2.2 的 1 档强调“核心动作有正面证据”，却在 2 档才突出“明确要求保持的动作分量”，可能使明确要求保持的朝向被当作只影响满档的细节。这个措辞风险与本次视觉漏检应分开记录，单次结果不足以证明因果。\n\n建议判定顺序：先在 BEFORE 锁定指定人物并定位其 AFTER 对应者；遇到数量保持要求，两图分别逐个清点、记录位置，再核查目标动作。新增者做出动作不能算原主体完成；明确要求保持的站位、朝向若发生实质变化，也不能自动降格为“非关键姿态细节”。本次仅记录建议，未改冻结 prompt。\n\n\n**sol 修订轮评分后复核：** 反馈回归部分改善：Gemini 总分 60→0，正确抓到明确锁定的背向被改变；但仍写“左侧1+货架5=6”，漏掉源图目标仍在而新增一人的实际 6→7，第二维 1 仍过高。动作分量约束起效，独立计数要求未被正确执行。', 'e010': '### e010 / Gemini：评分后复核（原始评分保留）\n\n复核结论：Gemini 图中没有找到原理由声称的果实数量变化、枝叶大范围重构或远景布局漂移。中央果串前后均为五颗，右下果串前后均为七颗；逐颗排列、青紫颜色分布与相邻枝叶关系能对应。叶片、海岸、岩壁和地面花环的主要几何也总体保持。局部纹理和清晰度有细微差异，不等于对象被增删或大幅移位。\n\n实际评分输入为 BEFORE 1792×2400、AFTER 896×1200，长宽恰好各缩为一半、宽高比相同。同分辨率局部对照见下图；六处保留区域的特征匹配点在配准修正前按画幅归一化计算，中位横向偏移绝对值小于画宽 0.022%，纵向不超过画高 0.015%。这支持主要布局保持，不是逐像素相同或每片叶子都已完整清点的证明。\n\n记录对齐：用户引用的“原始 1、60 分；叶片和果实数量、形状与位置出现多处重绘”来自 V2.0 / Gemini。V2.0 全部原始档位为 (1,1,1)、总分 60；V2.1 为 (0,0,2)、总分 0；V2.2 也是 (0,0,2)、总分 0，进一步声称果实数量、排列、颜色及海岸轮廓明显改变。后两版严重保持破坏的证据不成立；V2.0 的数量变化理由也不准确，但它还沿用“超常规”满档门槛，不能仅因撤回该理由就机械改为 V2.2 的分数。\n\nV2.2 确已优化此处：replace / Visual Naturalness 的 0 档明确写“‘整体重绘’本身不是证据”；Prompt Compliance 的 2 档还要求“文件像素尺寸不同不等于对象缩放；数量变化须指出具体新增或缺失对象”。因此本次不是规则明确允许按重绘感扣分，而是判官没有落实对应核验，并写出了不成立的具体差异。\n\n仍需修正的措辞位置：数量与几何核验应约束所有档位，尤其是准备以此判 0 或从 2 降至 1 时，不能只放在 2 档条件里。建议补为：“对任何保持扣分，先指出同一对象或固定参照下的具体前后差异；以数量变化扣分须定位新增或缺失对象。不能用重绘、重构、场景漂移等概括替代差异核验；无法确认的变化不得写作已发生的事实。已确认的细节损失再按其实际影响区分 1 与 0。”\n\n本次撤回对“大范围变化”的认可，保留原始冻结评分与总体均分；不将复核当作新的盲评分，也未改本轮冻结 prompt。\n\n![橄榄果串同尺度对照](../synth_v61_pilot/scores_qib_v22/audits/e010_olive_pixel_review_20260907.png)\n\n[数量与几何核验记录](../synth_v61_pilot/scores_qib_v22/audits/e010_geometry_review_20260907.json)\n\n\n**sol 修订轮评分后复核：** 反馈回归通过：Gemini raw (0,0,2)→(2,2,2)，总分 0→100。正确对应枝叶、果实分组和远景，没有再用“重绘”虚构数量或布局变化。此项是已知误报修复，不代表同类题均已解决。', 'e011': '### e011 / Gemini 与 Bagel：评分后复核（近景误判与实际越界分开核验）\n\n复核结论：用户引用的 V2.0 与 V2.2 两段均来自 Gemini。同尺度核验支持近景基本保持，原理由中的蘑菇数量及位置明显改变、工人移至中央并改变姿态、粘虫板换成另一套排列，没有成立的对应证据。不过，左侧及中部远处的多排稻草栽培床确被新农田与雾景替换；题面锁定“全部栽培床和稻草”，这处实际越界仍足以支持主维 0，不能因近景误判而直接改判通过。\n\nGemini 的 BEFORE 为 2752×1536，AFTER 为 1376×768，长宽恰各缩为一半，宽高比不变。三簇蘑菇的可见菌盖分别为左五、中央七、右五，前后能逐个对应。工人的帽子、弯腰躯干、手臂和手的位置均总体对应。工人区域有 48 个有效特征匹配，其中位位置差约为画宽 −0.012%、画高 −0.021%，不支持“从画面右侧移至更靠中央”。\n\n前方最大粘虫板有 31 个有效匹配，中位位置差约为画宽 −0.009%、画高 −0.022%；其吊点和主要轮廓对应，其他可见粘虫板也保持同一主要排列。三簇蘑菇分别有 80、112、87 个有效匹配，其相对画幅位置总体稳定。中央另一粘虫板仅有 1 个匹配，不据此单独作几何统计结论。匹配点偏移均在配准修正之前按各自画幅归一化计算，不能代替对全部纹理、文字或未匹配区域的审查。\n\n实际可定位的保持失败在诊断图第三行：以黄色粘虫板和立柱为固定参照，BEFORE 的粘虫板下方可见连续延伸的稻草床列，AFTER 对应区域变为绿色农田、树木和雾带，近处的一排床沿仍在。被替换的床列虽在画面远处，仍属于题面明确要求保留的栽培床；不能仅因位置远，就自动纳入允许修改的室外背景。\n\n建议将主维理由改为：“室外薄雾青山与农田已形成，近景三簇蘑菇、工人姿态及粘虫板主要布局保持；但左侧粘虫板下方的多排原有稻草栽培床被农田与雾景替换，违反全部栽培床和稻草不变的明确要求，因此主维为 0。”这段只保留可定位的越界依据，不再声称近景人物、蘑菇和构件全面移位。\n\nV2.2 的 background 主维 0 档已规定：须定位具体内容或越界位置，仅“整体重绘”不足以证明违规；透过透明围护看到的新背景不等于前景本体被改写。此例需要先把题面锁定的全部对象对应完整，再判编辑边界。V2.0 的二维护持失败理由也混入了主维越界判断；V2.2 已将融合与物理维度独立评价，本次不因主维 0 自动重定其余两维。\n\nBagel 候选另行核看：三簇蘑菇、工人及主要粘虫板的几何同样总体对应，不能泛称数量与姿态全部变化；但远处床列大面积模糊、被青蓝色区域替代，前景纹理显著软化，成片农田也缺少清楚证据。因此其失败仍有可定位依据，应与 Gemini 的近景保持情况分别描述。\n\n本次为评分后复核，只更正理由的证据解释，不产生新盲评分。Gemini 原始 V2.0 (0,0,1)、V2.1 (0,2,2)、V2.2 (0,1,1) 与各版总分 0 均保留；全部冻结评分、总体均分和本轮 prompt 未改。下图按相同坐标裁切，BEFORE 先降采样至 AFTER 分辨率，无锐化、超分或生成式补全。\n\n![e011 近景、工人及远处栽培床同尺度对照](../synth_v61_pilot/scores_qib_v22/audits/e011_foreground_pixel_review_20260907.png)\n\n[对应与几何核验记录](../synth_v61_pilot/scores_qib_v22/audits/e011_geometry_review_20260907.json)\n\n\n**sol 修订轮评分后复核：** 反馈回归未通过：Gemini 主维仍为 0，总分仍为 0，但理由仍虚构近景蘑菇、工人、粘虫板和骨架大幅几何变化。真正应抓的是远处锁定栽培床被农田/薄雾替代；本轮未给出这一对应证据。总分碰巧正确不能算事实核验正确。', 'e012': '### e012 / Gemini：评分后复核（确认 V2.2 第三维 1 → 0 的有效改进）\n\n复核结论（用户确认的有效改进）：Gemini 的 Physical & Detail Integrity 从 V2.0/V2.1 的原始 1 档降到 V2.2 的 0 档是合理的。原供水槽被移除后，水流仍从孤立的短出水构件流向碗中，可见上游供水连接没有成立；这是目标移除后留下的来源与结构矛盾，不能作为不影响可信性的轻微细节偏差给 1。\n\nV2.0 的理由实际已经写出“供水结构来源不清”，但又以“水流落入碗中的路径、人物姿势和三件尖头器具整体结构可辨且可用”支持 1 档，并把问题收束为“未达到超常精度”。它确认了水流下游落点与无关保留对象的结构，却没有对被编辑破坏的上游供水关系给予足够严重度。V2.1 同样发现短构件与细木桩的结构不可信，仍给了 1。\n\nV2.2 的有益变化是明确本维评价“目标原位置、补全内容及受移除影响的场景关系”，将“无来源的持续效果”列为物理失败的例子，并在 2 档提醒不能只查补全纹理而漏掉实际受影响的支撑、来源或光学关系。判断由局部画得像不像，扩展到移除之后相关物理关系能否成立；本例的 0 有对应图像依据，应保留这项规则边界。\n\n更稳妥的理由措辞：“原图木槽承接并输送水流；结果移除原上游槽体后，仍保留由孤立短出水构件流向碗中的水流，替代部件没有形成与保留场景连续的可辨供水连接，原供水关系被破坏，物理可信性不成立，因此记 0。”无需断言短构件究竟是石头、木头或完全封闭，也不额外假定隐藏管道。\n\n判据边界：应根据 BEFORE 已成立的供水关系、AFTER 可见的删除或断开，以及仍保留的依赖效果综合判断。不能把所有“水源没在画内”的场景都判 0，也不能仅因移除任务没完成就自动给物理维 0。本例有独立的物理关系反证；人物、碗和器具仍画得自然，不能抵消该矛盾。\n\n分数变化发生在 Gemini 第三维原始档位：V2.0=1、V2.1=1、V2.2=0。三版主维均为 0，因此第三维钳后都为 0，official 总分也均为 0。应把它记录为维度判断变准确的案例，不能因总体得分不变而漏掉这项改进。Bagel 的无源悬空水流更直接，第三维在三版中本来就都是 0。\n\n本条作为后续修订应保留的正向案例：纠正其他题里缺少证据的“重绘、移位”扣分时，仍需对可定位且破坏可信性的物理矛盾直接判 0。本次仅添加评分后复核注，冻结评分、总体均分和本轮 prompt 均未改。\n\n\n**sol 修订轮评分后复核：** 明确回退：Gemini 物理维由 0→2。新理由只核验短出水构件的支撑以及水流进入碗的下游关系，漏掉编辑删除原有可见供水通路、却保留持续出水的矛盾；主维仍为 0，所以总分掩盖了第三维回退。Bagel 无来源悬空水流仍判 0，不能抵消 Gemini 漏判。', 'e015': '### e015 / Gemini：评分后复核（局部铭牌失真，主要几何保持）\n\n复核结论：Gemini 的女跑者下肢、左侧跑步机控制台及扶手、透明跑步机与两名维修人员的主要姿态和布局总体对应，不支持 V2.2 理由中的“大幅重绘”“整体向左靠近”“机身比例与位置明显改变”。但不是所有细节完全一致：透明跑步机底部铭牌的部分字符存在可辨字形失真，因此第二维以局部细节损失给 1 比判结构保持失败的 0 更合适。\n\nBEFORE 为 2752×1536，AFTER 为 1376×768，长宽恰好各缩为一半，宽高比不变。七处已对应区域的匹配点，在配准修正之前按画幅归一化计算，中位横向位置差绝对值不超过画宽 0.013%，纵向不超过画高 0.019%。两名维修人员各有 137、196 个有效匹配，左控制台有 36 个，透明机器上部及底座分别有 164、111 个；这支持主要几何保持，不保证每个像素和小字相同。\n\n同分辨率局部对照可辨：左控制台的 HR 148、DIST 3.2、CAL 215 对应，维修人员的持工具、指向机身及托夹板姿势对应，工装 WNQ 标识主要保持。底部铭牌原有 ISO9001 及 WNQ 1989 的部分字符在 AFTER 出现笔画变形；应记录“铭牌局部字形失真”，不凭此猜定一套新的文字，更不能推断两名人员及整台机器移动。\n\n建议第二维理由：“女跑者的背向身份、黑色背心、灰色长裤及腿部步态总体保持；两名维修人员的姿态、工具与夹板关系，以及两台跑步机的主要结构和布局对应。透明跑步机底部铭牌有局部字形失真，属于未破坏主要身份与结构的细节保持偏差，因此给 1。”V2.0 的 1 档可由这项实际细节支持，其原理由“全画面轻微重绘”仍过于笼统。\n\n第一维的动作失败仍有独立依据：AFTER 双臂主要向左右展开，未完成沿头部两侧斜向前上方举过头顶。因此本次第二维纠偏不改变 official 总分 0；第三维的独立解剖判断亦不因第二维修正而机械改动。\n\n反复误判暴露的共性：V2.2 虽已要求不要凭“重绘”扣分，但数量与几何核验还分散在若干 2 档条款里，JSON 示例也把分数放在 observations 之前。修订草案将证据要求前置到所有档位，先写“对象位置、BEFORE 状态、AFTER 状态、影响要求”，再落档；数量分图清点，文字区分具体改字与可读性退化，且不把局部纹理或字形损失扩大成整体语义与布局变化。顺序调整是否提高准确性仍需复评验证，不能仅凭文本更完整就宣布有效。\n\n草案同时补足 action 的明确保持条件：通过档也需核验锁定的动作分量，避免 e006 的错主体和朝向要求被弱化；remove 的物理判据保持不变，保留 e012 的无来源水流判 0。e011 提醒对远处锁定对象同样逐项对应，不能只因近景保持就放过远处床列丢失。\n\n输入呈现建议另由判分编排实现：两图使用同一显示尺度，并保留原始像素供对应区域局部放大；不能为统一尺度而永久丢弃原图的小字证据。显示大小差异可能诱发错配，但本次未做控制实验，不能把全部误判都归因于分辨率。当前仅输出 prompt 修订草案及 diff，未替换冻结 V2.2、未重跑判分；草案已通过九类型加载与二十题渲染检查，尚未验证打分准确率。\n\n[证据流程修订 diff 草案](../synth_v61_pilot/qib_v22_evidence_revision_proposal_20260907.diff) · [完整 prompt 草案](../synth_v61_pilot/qib_v22_evidence_revision_draft_20260907.md)\n\n![e015 器械、维修人员及铭牌同尺度对照](../synth_v61_pilot/scores_qib_v22/audits/e015_preservation_pixel_review_20260907.png)\n\n[对应与几何核验记录](../synth_v61_pilot/scores_qib_v22/audits/e015_geometry_review_20260907.json)\n\n\n**sol 修订轮评分后复核：** 反馈回归部分改善：Gemini 第二维 0→1，动作失败仍为 0；但新理由仍以器械局部比例、支架轮廓和位置变化扣分，未定位同尺度审计确认的铭牌字形损失。严重度改善，扣分证据未完全纠正。', 'e019': '### e019 / Gemini：改前 V2.2 评分后复核（非目标几何保持）\n\n复核结论：不支持改前冻结 V2.2 理由中的“右侧检测仪及注射笔明显向中央移近”“玻璃杯、试管架和绷带卷发生位置与尺度变化”。这些非目标的形状、主要布局和前后对应都成立；此项扣分应撤销，第一维按可见替换及保持证据可支持 2。该结论是评分后审计，旧 raw (1,2,2) 与总分 60 原样保留。\n\nBEFORE 为 1448×1086，AFTER 为 1200×896。未做配准修正前，已匹配的右侧检测仪、注射笔、玻璃杯、试管架和绷带卷，其匹配点归一化位置的中位横向差绝对值均小于画宽 0.014%，纵向差均小于画高 0.113%。这与大幅移动的说法不符；指标仅辅助判断已对应区域的几何，不证明每个纹理像素完全相同。\n\n中央深蓝检测仪及白色试纸被完整替换为直立透明玻璃小瓶，灰色橡胶塞、银色压盖、约半瓶无色液体和无文字标签均有可見支持。位置比较应以画幅和固定物体的对应关系为准；不同文件尺寸、轻微锐度差或笼统“重绘”不能代替实际位移证据。\n\n[几何核验记录](../synth_v61_pilot/scores_qib_v22/audits/e019_geometry_review_20260907.json)\n\n\n**sol 修订轮评分后复核：** 反馈回归未完全通过：Gemini 仍 raw (1,2,2)、总分 60。新理由撤回“其他物体明显向中央移动”，却改以轻微画幅与边距差异降档。归一化对应不支持对象位移或有验收意义的构图损坏，不能将换一种说法视为已修复；第一维仍建议按已核实事实给 2，冻结分数不改。', 'e001': '\n\n**sol 修订轮评分后复核：** 额外复核：Gemini 主维 2→1 的下降有可见依据。按相同显示宽度并以相邻红色机器人为参照，替代雕塑的总高度高于原橙色屈膝机器人，未精确满足“等高”；这属于连续量执行偏差，不是本次误报回退。'}

@lru_cache(maxsize=80)
def _review_figure(path, title):
    path = Path(path)
    with Image.open(path) as source:
        size = source.size
        preview = source.convert('RGB')
        preview.thumbnail((720, 720), Image.Resampling.LANCZOS)
        buf = io.BytesIO()
        preview.save(buf, format='JPEG', quality=85)
    uri = 'data:image/jpeg;base64,' + base64.b64encode(buf.getvalue()).decode()
    caption = f'{title} · {size[0]} × {size[1]}'
    return ('<figure style="flex:1;min-width:240px;margin:6px 0">'
            f'<img src="{uri}" alt="{_review_esc(title)}" style="width:100%;height:auto;max-width:720px">'
            '<figcaption>' + _review_link(path, caption + '（原始 PNG）') + '</figcaption></figure>')

def _review_judge(candidate, version):
    p = REVIEW_JUDGES[candidate, version]
    if not p['actual_configurations']:
        return '登记 sol；运行日志未找到'
    return '、'.join(model.removeprefix('gpt-') + ' / ' + effort for model, effort in p['actual_configurations'])

def _review_scores(candidate, qid):
    versions = [(v, folder) for v, folder in REVIEW_RUNS if (candidate, v) in REVIEW_SCORES]
    parts = ['<table class="qib-merged-table"><thead><tr><th>版本</th><th>判官 / 推理档位</th><th>总分</th><th>原始三维 → 钳后三维</th></tr></thead><tbody>']
    for version, _ in versions:
        r = REVIEW_SCORES[candidate, version][qid]
        raw = tuple(d['tier'] for d in r['raw_dimensions'])
        clipped = tuple(r['official_tiers'][f'd{i}'] for i in range(1,4))
        parts.append(f'<tr><td>{_review_esc(version)}</td><td>{_review_esc(_review_judge(candidate,version))}</td><td><b>{_review_num(r["official_total"])}</b></td><td>{raw} → {clipped}</td></tr>')
    parts.append('</tbody></table>')
    for i in range(3):
        label = REVIEW_SCORES[candidate, versions[0][0]][qid]['raw_dimensions'][i]['label']
        parts += [f'<h4>维度 {i+1} · {_review_esc(label)}</h4>',
                  '<table class="qib-merged-table"><thead><tr><th>版本</th><th>原始 → 钳后（分数）</th><th>判官理由原文</th></tr></thead><tbody>']
        for version, _ in versions:
            r = REVIEW_SCORES[candidate, version][qid]
            d = r['raw_dimensions'][i]
            k = f'd{i+1}'
            mark = f'{d["tier"]}（{d["mapped"]} 分） → {r["official_tiers"][k]}（{r["official_dimensions"][k]} 分）'
            parts.append(f'<tr data-candidate="{candidate}" data-version="{_review_esc(version)}" data-dim="{k}"><td>{_review_esc(version)}</td><td>{mark}</td><td class="judge-reason">{_review_esc(d["reason"])}</td></tr>')
        parts.append('</tbody></table>')
    return ''.join(parts)

def _review_legacy(qid):
    parts = []
    for name, values in REVIEW_LEGACY.items():
        r = values.get(qid)
        if not r: continue
        parts.append(f'<h5>{_review_esc(name)} · {_review_num(r["official_total"])} / 5</h5>')
        for d in r['raw_dimensions']:
            parts.append(f'<p><b>{_review_esc(d["label"])}：{d["score"]} / 5</b> · {_review_esc(d["reason"])}</p>')
    return ''.join(parts)

def _review_case(qid):
    m = REVIEW_INPUTS['Gemini'][qid]
    parts = [f'<article class="qib-merged-case" data-qid="{qid}">',
             f'<h3>{qid} · {_review_esc(m["edit_type"])}</h3>',
             f'<p><b>题面：</b>{_review_esc(m["edit_instruction"])}</p>',
             '<div style="display:flex;gap:16px;flex-wrap:wrap;align-items:flex-start">',
             _review_figure(m['before'], 'BEFORE 原图'),
             _review_figure(m['after'], 'Gemini AFTER'), '</div>',
             '<h4>Gemini · 各版与判官配置评分</h4>',
             _review_scores('Gemini', qid)]
    review = ''
    if qid in REVIEW_AUDIT:
        review += '<p><b>最近一次复核：</b>' + _review_esc(REVIEW_AUDIT[qid]) + '</p>'
    if qid in EARLY_REVIEW_NOTES:
        review += _review_fold('早期像素对照与复核记录（保留当时表述）', _review_markdown(EARLY_REVIEW_NOTES[qid]))
    if review:
        parts.append(_review_fold('复核说明与像素证据（评分后记录）', review))
    bm = REVIEW_INPUTS['Bagel'][qid]
    bagel = _review_figure(bm['after'], 'Bagel AFTER') + _review_scores('Bagel', qid)
    parts.append(_review_fold('Bagel 图像与四版评分', bagel))
    parts.append(_review_fold('其他口径：官方 1–5 与旧 1–5 审计', _review_legacy(qid)))
    links = []
    for version, folder in [REVIEW_RUNS[1], REVIEW_RUNS[2], REVIEW_RUNS[3]]:
        prompt = REVIEW_BASE / folder / 'gemini/prompts' / f'{qid}.txt'
        links.append(_review_link(prompt, version + ' prompt'))
    parts.append('<p><b>本题规则：</b>' + ' · '.join(links) + '。相同类型的规则全文统一放在下方附录。</p>')
    if qid in globals().get('QS', {}):
        q = QS[qid]
        meta = {k:q[k] for k in ['targeting_types','consequence_types','preservation_types','scene_types','hop_types','knowledge_categories','weak_points','level_reason'] if q.get(k)}
        if meta:
            parts.append(_review_fold('出题要素', '<pre>' + _review_esc(json.dumps(meta,ensure_ascii=False,indent=2)) + '</pre>'))
    parts.append('</article>')
    return REVIEW_STYLE + ''.join(parts)

REVIEW_STYLE = '<style>.qib-merged-case{line-height:1.6}.qib-merged-table{width:100%;border-collapse:collapse;margin:10px 0}.qib-merged-table td,.qib-merged-table th{border:1px solid #8885;padding:7px;vertical-align:top;text-align:left}.qib-merged-table td:first-child{min-width:100px}.qib-merged-case details{margin:10px 0}.qib-merged-case summary{cursor:pointer;font-weight:600}.qib-merged-case pre{white-space:pre-wrap}.qib-merged-case figcaption{font-size:12px}</style>'
for _qid in REVIEW_QIDS:
    _html = _review_case(_qid)
    # Plain-text fallback contains the same content once, with tables laid out as text.
    from bs4 import BeautifulSoup
    _plain = BeautifulSoup(_html, 'html.parser').get_text('\n', strip=True)
    display({'text/html': _html, 'text/plain': _plain}, raw=True)

# 规则按编辑类型去重：V2.0 归档只列一份，其他三版每类各列一份。
_archive = REVIEW_BASE.parent / 'archive/judge_prompt_edit_qib_v2.md'
_parts = ['<section class="qib-merged-case"><h3>评分规则全文（按类型合并）</h3>',
          _review_fold('V2.0 归档规则（阅读对照；非历史轮完整输入）', '<pre>' + _review_esc(_archive.read_text()) + '</pre>')]
_by_type = {}
for _qid in REVIEW_QIDS:
    _by_type.setdefault(REVIEW_INPUTS['Gemini'][_qid]['edit_type'], _qid)
for _type, _qid in sorted(_by_type.items()):
    _body = []
    for _version, _folder in REVIEW_RUNS[1:4]:
        _prompt = (REVIEW_BASE / _folder / 'gemini/prompts' / f'{_qid}.txt').read_text()
        _rules = _prompt.split('# 二、打分规则',1)[1].split('# 三、输出',1)[0]
        _rules = _rules.split('\n',1)[1].strip()
        _body.append(_review_fold(_version, '<pre>' + _review_esc(_rules) + '</pre>'))
    _parts.append(_review_fold(_type + ' · V2.1 / V2.2 / V2.2 修订', ''.join(_body)))
_parts.append('</section>')
display({'text/html': ''.join(_parts), 'text/plain': '按九类编辑类型合并的评分规则；展开查看。'}, raw=True)
````

## 八、QIB v2 / v2.1 逐题复核：哪些是细则问题，哪些是判官误读（2026-09-07）

**核心判断：v2.1 的主要问题不是分数更低，而是“失败证据—严重度—维度归属”没有闭合。** 它在一些题上把没有充分证据的非目标变化直接升级为主维失败；在另一些题上，又把未完成动作当成可接受的不清晰。三维钳制把这些主维不稳定放大成整题归零或通过。与此同时，取消“超常规”门槛、将风格与内容保持分开，都是应保留的改进。

HTML 副本已移除；20 题图像与逐维评分继续在本 notebook 查看。原始数据：`../synth_v61_pilot/questions.jsonl`、`scores_qib/{gemini,bagel}/scores.jsonl`、`scores_qib_v21/{gemini,bagel}/scores.jsonl`。各题保留原图链接，点击图像可打开实际判官输入的 PNG。

**先看分数变化发生在哪里。** 下表是同一批图在显式标注的不同版本下如何被计分，不能解释为生成模型能力变化。

| 模型 | v2 official | v2.1 official | v2 未钳制 | v2.1 未钳制 | v2 主维 0/1/2 题数 | v2.1 主维 0/1/2 题数 |

| Gemini | 42.000 | 24.667 | 54.667 | 56.667 | 10 / 2 / 8 | 14 / 2 / 4 |

Gemini 的未钳制均分 **上升 2.00**，钳制损失却从 **12.667 扩大到 32.000**；official 的下降集中在主维。Gemini 新归零的是 **e003、e004、e010、e019**；Bagel 新归零的是 **e004、e008、e010、e013**，从零升出的则是 **e006、e012**。40 个候选的零分数从 24 增到 30，v2.1 两模型同为零的题达到 12/20。30 个零分候选中，有 23 个仍至少一项 raw d2/d3 非零，因此“总分 0”不能读作“三维都不成立”。

官方汇总机制始终为 `mean(φ(d1), φ(min(d2,d1)), φ(min(d3,d1)))`，`φ={0:0,1:60,2:100}`。例如 e001/G 的 raw 由 `(2,2,2)` 变成 `(1,2,2)`，未钳制只从 100 降到 86.667，official 却降到 60。**钳制是已有机制，不是 v2.1 新增的文字缺陷；但在这个机制下，主维 0/1 边界尤其不能含糊。**

**最优先修的五处文字问题**

1. **“任何变化”与“使任务不成立的变化”没有分开，0/1/2 也不完整。**

   replace 的 0 包含“数量、位置、关键属性任一不符”，1 又容许“色调、材质、朝向、尺度”的次要偏差；adjust 的 0 是“改变扩散到未被指定的对象”，1 只写目标内的程度或范围偏差；background 的 0 是“任一前景物体被重绘”；action 的保持维甚至把“服饰或配件改变”直接放进 0，未限定是否为身份或关键内容的实质改变。

   e001/G 的“等高略偏高”被接纳为 1，e010 两模型却因尚未核实的“体量偏小”得到 0。e004 中真实的局部车牌/污迹变化被判成 d1=0，e018/G 的非目标细节变化又只降到 1。规则还反复写“仅一处”偏差，未覆盖两处轻微但总体仍成立的情况。

   remove 的漏洞更直接：0 只列“未移除、主体残留、误删其他对象”，e003/G 的 0 理由却是其他物体移位。无论这项观察真不真实，现有档位没有明确承接这个理由。**不满足 2 的充分条件，不等于满足 0。**

   建议按各 TYPE 明确：核心操作/离散约束错、关键语义丢失、可定位的严重保持破坏进入 0；核心成立而连续量或非关键细节存在局部偏差进入 1；全部适用要求有正面证据进入 2。“一处”改成“局部偏差，累计不足以改变任务结论”。若业务要求逐像素零容忍，必须明确采用该目标并统一验收；不能同时写允许轻微局部变化，又把任意差异一票否决。

2. **以“重绘、缩放、移位”判失败，却没有要求判官指出可复核的对应差异。**

   e010/G、B 与 e019/G 的低分理由都声称非目标物大范围重排；同尺度局部图并不支持。e003/G 的自卸车后部因中央机器移除而暴露，也不能仅凭可见范围变化认定车移动。e004 的车牌和污迹可以具体指出，而所谓全幅构图漂移则不成立。

   本次只作辅助核验：把图像按画幅坐标归一化，做 SIFT 对应点与 RANSAC 主导相似变换估计。e004/G 有 1315 个内点，尺度系数约 1.0000；e010/G 有 937 个内点，约 1.0000；e019/G 有 571 个内点，约 1.0013。对应点位移中位数约为归一化坐标单位的 0.050%、0.047%、0.107%。这些结果支持“主要布局稳定”，**不证明每个像素或所有细节保持**，也不能单独决定评分。

   原图与输出的分辨率确实不同，例如 e004 为 2752×1536 → 1376×768，e010 为 1792×2400 → 896×1200，都是两轴各缩至一半。分辨率差异是误读的一种可能来源，**目前无法证明它就是判官误读的原因**。应把“同显示尺度、归一化位置、具体对象锚点”写入涉及保持的类型判据；0 分理由需写明哪个对象、原来怎样、现在怎样、对应哪条失败条件。

3. **没有规定通过档需要核心动作的正面证据，反而允许用不清楚来解释成功。**

   e006/B 是最明确反例：v2 判 0；v2.1 判 1，理由说交叠于上身前方，只是 X 的交叉点不够清楚。局部放大仍显示右手向上触碰货架，没有所需的双臂收回与交叉。这不是“动作已完成、细节略糊”，而是核心变化没有成立。主维这一误判将该候选总分从 0 变成 60。

**v2 旧文**：

> 指令指定的数量、位置、关键属性任一不符

来源：现行 v2.1，L54。
