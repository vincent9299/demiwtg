## 参考知识与图片角色

| 条件／角色 | 含义 |
| --- | --- |
| baseline｜无额外资料 | 仍有任务文本；edit仍有编辑原图。不是模型没有预训练知识。 |
| text｜文本 | 只加文字知识，不加知识图片。 |
| image｜图片 | 只加参考图片及必要角色说明，不加知识正文；仍保留任务文本。 |
| multimodal｜文本＋图片 | 同时加文字和参考图；有图不代表图已支持考点，更不代表图文互补已成立。 |
| edit_source｜编辑原图 | 待修改场景，两组共同输入；不是额外检索知识。 |
| retrieval_reference｜知识参考图 | 用于提供形态等证据；不是本题目标图，也不要求照抄构图。 |
