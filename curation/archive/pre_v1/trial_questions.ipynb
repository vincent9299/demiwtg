# 三道试验题：先看题目是否值得试

**直接阅读，不需要运行 cell。** 本页是助手准备的题目，没有替你打标，也没有调用生成模型。三题涉及两个知识点；螺栓的文生图与编辑是同一考点在两种任务下的比较，不是独立覆盖量。

这轮只验证“前提能否形成可观察、不会误伤合理画法的约束”，不声称满足旧出题 prompt 的完整难度、组合和跨域门槛。知识归因仍需后续对照实验；一张图画错，不能直接证明模型不知道知识。

每题先读“给模型的题面”，再看“审核者的判据”。正式生成只能发送题面（编辑题另附原图），不能把答案与判据一并发送。

## 统一怎么记录结果

每条知识判据分别记：**符合／明确冲突／无法观察**，附画面位置和一句依据。另记指令执行问题，例如缺了要求的对象、遮挡了指定部位。

- 关键部位看不清：不猜测，记无法观察；若题面要求清楚入画，同时记指令问题。
- 画得清楚且关系与知识冲突：记明确冲突，暂不推断模型内部原因。
- 只满足部分判据：逐项保留，不能只用“整体看起来对”通过。
- 符合全部知识判据但违反数量或保留要求：知识表现与指令执行分开记录。

以下“正确／错误例子”是文字边界测试，不是已经生成或经过实验验证的图片。


## T01．鸡腿菇正在自溶

**用途：文生图；状态：待你审阅题意，尚未试生成。**

**给模型的题面：**

> 画一幅鸡腿菇（Coprinus comatus）的自然观察特写，表现正在自溶、菌盖尚未完全消失的阶段。画面只以一株为主体，菌盖及其下缘清楚入画，不用文字说明。

**已给定的前提：** 指定物种；正在自溶但非完全消失；单株主体；菌盖下缘可见。

**仍需要模型自己知道：** 该物种此阶段菌盖逐渐解体，伴有墨汁样深色液化物。

**为什么没有直接给答案：** “自溶”是已知阶段，不计作推理；具体颜色和菌盖下缘形态没有在题面逐项给出。阶段词本身有提示作用，此题不宣称高难度。

### 审核者的判据（不发给生成模型）

- **K1**：菌盖仍有残余结构，且下缘可见局部消解、失去完整边缘的表现。
- **K2**：消解部位伴有墨汁样深色液化物；不能只画成完整白色菌盖旁的一团无关黑色背景。

**另外检查指令：** 单株主体；菌盖下缘可观察；没有以文字替代画面。

**允许的其他画法：** 可处于不同消解比例；可以附着液体而没有悬空液滴；不要求整株发黑，不限定滴数、背景或精确尺寸。

**边界测试：**

- 残余菌盖与下缘深色液化物相连，没有悬空水滴：可以符合。
- 完整干燥菌盖，没有消解迹象：K1 明确冲突；不能仅因旁边有黑斑就通过 K2。
- 菌盖下缘被草遮住：对应项无法观察，同时记录可见性指令问题。

**不据此判分：** 食用性、固定经过几小时、黑色像素比例、不可见微观过程。

**依据：** 资料说明菌盖消解及墨汁样物质，并配有约半个菌盖已消解的照片；据此拟定可见判据，不要求复制照片形状。

- [Australian National Botanic Gardens：Spore discharge and dispersal in mushrooms](https://www.anbg.gov.au/fungi/spore-discharge-mushrooms.html)，定位：The Inkcap mushrooms 小节；菌盖约消解一半的照片说明及后续过程解释。

**局限：** 仅验证这两个阶段表现，不构成完整物种鉴定；天然个体差异的判分宽容度仍需试生成与人工校准。

## T02．贯穿螺栓连接两块板

**用途：文生图；状态：待你审阅题意，尚未试生成。**

**给模型的题面：**

> 画一幅机械教学局部剖视图：两块平行金属板搭接，用一套普通六角头贯穿螺栓和六角螺母连接，处于完成夹紧的状态。剖切经过螺栓轴线，完整清晰展示连接处，不加文字或尺寸标注。

**已给定的前提：** 两板搭接；普通贯穿螺栓与螺母；已夹紧；经过轴线的局部剖视。

**仍需要模型自己知道：** 杆件穿过两板；头部和螺母在板件组合相对外侧形成夹持，螺母与杆件螺纹段配合。

**为什么没有直接给答案：** 指定连接类型与状态，没有直接列出零件排列顺序。图像表现只能验证装配关系，不能验证真实预紧力。

### 审核者的判据（不发给生成模型）

- **K1**：杆件经对应孔连续贯穿两块板，而不是只插入其中一块或穿过无孔实体。
- **K2**：头部与螺母位于两板组合的相对外侧；可经垫圈接触，形成夹持布置。
- **K3**：螺母套接在杆件螺纹段上，未悬空或与杆件分离；允许教学剖视的简化螺纹画法。

**另外检查指令：** 两块板、一套连接件；连接部位完整可见；采用要求的剖视表现。

**允许的其他画法：** 从左右任一侧穿入；可有垫圈；全螺纹和部分螺纹均可；可旋转整幅图。不规定螺纹圈数或露出长度。

**边界测试：**

- 方向与参考相反但穿接夹持一致：全部可以符合。
- 两板都画出，但杆件只穿过第一块：K1 明确冲突。
- 头部、螺母同在板件组合一侧且另一侧没有夹持端：K2 明确冲突。
- 内部被完全遮挡：K1 无法观察，另记未按剖视要求呈现。

**不据此判分：** 真实夹紧力、摩擦力数值、强度安全、细微螺纹制图规范；不把全螺纹判错。

**依据：** 作者解释紧固件与被夹件的配合，并展示螺栓螺母固定支架与支承板。这里把结构关系用于两板教学题，是基于资料的设计推断；不要求从图推断载荷。

- [Bill Eccles / Bolt Science：The basics of bolted joints](https://www.boltscience.com/pages/the-basics-of-bolted-joints.pdf)，定位：第 1 页正文与 Figures 3、5；第 2 页关于板件夹紧的说明。

**局限：** 这是装配空间关系试题，不评估工程设计是否安全；剖视简化导致看不清时需保留无法观察。

## E01．把原图左起第一组零件装成连接

**用途：图像编辑；状态：待你审阅题意，尚未试生成。**

**给模型的题面：**

**编辑题的原图：** 左起第一组是目标，另外四组为保留对象。图片作为编辑输入，不作为装配答案。

## 接下来由你看什么

只需要看三段题面，反馈“可以试”或者指出具体哪题不符合目标。无需重新审核此前全部知识卡。

题意确定后，建议首轮每题生成 2 张，共 6 张（同一题保留全部结果，不挑最好的一张）。两张只用于暴露判据问题，不足以估计模型准确率。编辑题使用同一张原图。应先固定题面和判据版本，再看生成结果；如需改题，记录新版本，不倒改标准让旧结果通过。
