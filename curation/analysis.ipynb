# IP 实例数据分析（DuckDB + JupySQL）

数据源：`data/taxonomy/instances.json`（相对仓库根）（59,687 条实例）

- `%%sql` 单元格：直接写 SQL
- 结果可赋值给变量转 pandas：`df = %sql SELECT ...`

```python
%load_ext sql
%config SqlMagic.autopandas = True   # 查询结果直接返回 DataFrame
%config SqlMagic.feedback = False
%sql duckdb:///:memory:
```

```python
%%sql
-- 展开 instances 数组为可查询的视图（文件 25MB，需调大 maximum_object_size）
CREATE OR REPLACE VIEW inst AS
SELECT i.name AS name, i.taxonomy_path AS taxonomy_path,
       i.source AS source, i.desc AS descr,
       i.aliases AS aliases, i.query AS query
FROM (SELECT unnest(instances) AS i
      FROM read_json_auto('../data/taxonomy/instances.json', maximum_object_size=1073741824));
```

```python
%%sql
-- 按二级分类统计（一级都是"融合世界标签体系"）
SELECT *
FROM inst
limit 10
```

```python
%%sql df_alias <<
-- 别名反查：别名 -> 正式名
SELECT i.name, alias
FROM inst, unnest(aliases) AS t(alias)
WHERE alias ILIKE '%机器猫%';
```

```python
# SQL 结果转 pandas 继续分析
df_alias.head()
```

```python
# 画图示例：二级分类分布
import matplotlib
matplotlib.rcParams['font.sans-serif'] = ['Noto Sans CJK SC', 'WenQuanYi Zen Hei', 'SimHei', 'DejaVu Sans']
matplotlib.rcParams['axes.unicode_minus'] = False

df = %sql SELECT split_part(taxonomy_path, '/', 2) AS cat, count(*) AS n FROM inst GROUP BY 1 ORDER BY n DESC
df.plot.bar(x='cat', y='n', figsize=(12, 5), title='二级分类实例数分布');
```

```python
%%sql
-- 常用：全文关键词搜索（名称/描述/别名）
SELECT name, taxonomy_path, descr
FROM inst
WHERE name LIKE '%迪士尼%'
   OR descr LIKE '%迪士尼%'
   OR list_contains(aliases, '迪士尼')
LIMIT 20;
```
