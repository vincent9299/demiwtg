````python
# 1. 直接读 Lance 表：题目、图对、参考和耗时（只读固定版本）
from pathlib import Path
import json, sys
PROJECT = Path("/yzp/zhaozy/yangzepeng/0905/demiwtg")
sys.path.insert(0, str(PROJECT))
from demiflow import data
from demiflow.lance.blobs import BlobRef
from IPython.display import display, Markdown, Image
from project import resolve_root

RUN_ID = "qwen21_pilot_20260923_02"  # 一次完整运行，包含两个 case
TABLE_URI = resolve_root() / "demiwtg/curation/edit/datasets/questions_1f69373914f05bb7b96b__qwen21_pilot_20260923_02.lance"
VERSION = 1
CONCEPT = ""  # 空=全部；也可填 莜面栲栳栳 或 绞胎瓷
rows = data.read_lance(str(TABLE_URI), version=VERSION).filter(lambda r: not CONCEPT or r["concept"] == CONCEPT).take_all()
display(Markdown("机器审核结论保留原样：莜面案例的训练指令与目标矛盾，审核文字还混淆了目标图与依据图的编号；请结合下面明确标注的原图/目标查看。"))
for row in rows:
    detail = json.loads(row["details"])
    display(Markdown(f"### {row['concept']} · {row['status']}"))
    for label, key in [("训练指令", "instruction"), ("学习目标", "learning_objective"), ("编辑类型", "edit_type"), ("已有图角色", "existing_role"), ("角色理由", "role_reason"), ("合成指令", "synthesis_instruction"), ("审核结论", "reason")]:
        display(Markdown(f"**{label}：** {row[key]}"))
    print({key: row[key] for key in ["design_seconds", "synthesis_seconds", "review_seconds", "model_load_seconds"]})
    for label, key in [("已有图片", "existing"), ("训练原图", "source"), ("监督目标", "target")]:
        if row[key]:
            display(Markdown(f"**{label}**"))
            display(Image(data=BlobRef(**row[key]).read(resolve_root()), width=600))
    for item in (row["input_content"] or [])[2:]:
        display(Markdown("**实际作答参考**"))
        display(Image(data=BlobRef(**item["blob_ref"]).read(resolve_root()), width=500) if item["type"] == "image" else Markdown(item["text"]))
    display(Markdown("**逐项审核**\n```json\n" + json.dumps(detail.get("review", {}), ensure_ascii=False, indent=2) + "\n```"))
````
